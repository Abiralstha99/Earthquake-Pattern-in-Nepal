```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data_path = Path("../data/processed/cleaned.csv")

df = pd.read_csv(data_path)
```

## Earthquake magnitude and depth - Are magnitude and depth related?

This analysis looks at 1,422 earthquakes to see whether deeper earthquakes tend to have higher or lower magnitudes. We used a scatterplot because depth and magnitude are both numerical, and each point shows the two measurements for one earthquake.

```python
# Calculate Pearson correlation
correlation = df["depth"].corr(df["mag"])
print(f"Pearson r = {correlation:.3f}")

# Create scatterplot
plt.scatter(df["depth"], df["mag"], alpha=0.3, s=20)
plt.xlabel("Depth (km)")
plt.ylabel("Magnitude")
plt.title("Earthquake Magnitude vs. Depth")
plt.show()

# Key findings
print(df.loc[df["mag"].idxmax(), ["depth", "mag"]])
print(df.loc[df["depth"].idxmax(), ["depth", "mag"]])
```

The scatterplot shows no clear upward or downward trend. The strongest clusters are vertical bands at recorded depths of 10 and 33 km, with a range of magnitudes at both depths. Two events stand out for different reasons. The largest earthquake had a magnitude of 7.8 at a depth of 8.22 km, while the deepest earthquake was 143.2 km deep and had a magnitude of 4.3.

Pearson’s correlation was −0.016. The negative sign points to a slight downward relationship, but the value is so close to zero that there is almost no linear association between depth and magnitude in this catalog.

## Magnitude distribution

```python
plt.hist(df["mag"], bins=20, edgecolor="white")
plt.xlabel("Magnitude")
plt.ylabel("Number of earthquakes")
plt.title("Earthquake Magnitude Distribution")
plt.show()
```

We used a histogram to show how many earthquakes fell into each magnitude range. About 77% of events had magnitudes from 4.0 up to but below 5.0. The tallest bar, roughly 4.20 to 4.44, contained 442 events. The distribution is right-skewed, with a long tail toward higher magnitudes and only 13 events at magnitude 6.0 or above.

## Conclusion

Overall, earthquakes at the same recorded depth varied in magnitude, and the correlation was nearly zero. Depth alone appears to be a poor predictor of magnitude in this catalog. These results describe the recorded events and do not establish cause and effect.
